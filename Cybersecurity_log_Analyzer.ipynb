```python
logs = """2026-08-22 10:15:22 | admin |
LOGIN_SUCCESS | 192.168.1.10
2026-08-22 10:16:05 | vivek |
LOGIN_SUCCESS | 192.168.1.15
2026-08-22 10:20:11 | admin |
LOGIN_FAILED | 192.168.1.50
2026-08-22 10:20:15 | admin |
LOGIN_FAILED | 192.168.1.50
2026-08-22 10:20:19 | admin |
LOGIN_FAILED | 192.168.1.50"""

for log in logs.splitlines():
  print(log)
```

```python
failed_count = 0

for log in logs.splitlines():
  if "LOGIN_FAILED" in log:
    failed_count += 1

print("Total Failed Login Attempts:",
      failed_count)
```

```python
failed_ips = {}

for log in logs.splitlines():
  if "LOGIN_FAILED" in log:
    parts = log.split("|")
    ip = parts[-1].strip()

    if ip in failed_ips:
      failed_ips[ip] += 1
    else:
      failed_ips[ip] = 1

print("Failed Login Attempts by IP:")
print()

for ip, count in failed_ips.items():
  print(ip, "-", count, "failed attempts")
```

```python
threshold = 3

print("Security Analysis")
print("-----------------")

for ip, count in failed_ips.items():
  if count >= threshold:
    print("^ SUSPICIOUS IP:", ip)
    print("Failed Attempts:", count)
  else:
    print("~ Normal IP:", ip)
```

```python
logs = """2026-08-22 10:15:22 | admin |
LOGIN_SUCCESS | 192.168.1.10
2026-08-22 10:16:05 | vivek |
LOGIN_SUCCESS | 192.168.1.15
2026-08-22 10:20:11 | admin |
LOGIN_FAILED | 192.168.1.50
2026-08-22 10:20:15 | admin |
LOGIN_FAILED | 192.168.1.50
2026-08-22 10:20:19 | admin |
LOGIN_FAILED | 192.168.1.50
2026-08-22 10:25:10 | user1 |
LOGIN_FAILED | 192.168.1.25
2026-08-22 10:30:10 | admin |
LOGIN_FAILED | 192.168.1.100
2026-08-22 10:30:15 | admin |
LOGIN_FAILED | 192.168.1.100
2026-08-22 10:30:20 | admin |
LOGIN_FAILED | 192.168.1.100
2026-08-22 10:30:25 | admin |
LOGIN_FAILED | 192.168.1.100"""

#count failed login attempts by IP

failed_ips = {}

for log in logs.splitlines():
  if "LOGIN_FAILED" in log:
    ip = log.split("|")[-1].strip()
    failed_ips[ip] = failed_ips.get(ip, 0) + 1

#security report

total_logs = len(logs.splitlines())
successful_logins = sum ("LOGIN_SUCCESS"  in log for log in logs.splitlines())
failed_logins = sum ("LOGIN_FAILED" in log for log in logs.splitlines())


threshold = 3
suspicious_ips = sum (count >= threshold for count in failed_ips.values())
normal_ips = sum (count < threshold for count in failed_ips.values())


print("========== SECURITY REPORT ==========")
print("Total Log Entries", total_logs)
print("Successful Logins", successful_logins)
print("Failed Logins", failed_logins)
print("Suspicious IPs", suspicious_ips)
print("Normal IPs", normal_ips)
print("=====================================")
```

```python
import matplotlib.pyplot as plt

#data
categories = [
    "Success Logins",
    "Failed Logins",
    "Suspicious IPs",
    "Normal IPs"
]

values = [
    successful_logins,
    failed_logins,
    suspicious_ips,
    normal_ips
]

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(categories, values)

plt.title("Cybersecurity Log Analysis")
plt.xlabel("Security Events")
plt.ylabel("Count")

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
