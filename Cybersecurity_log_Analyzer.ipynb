```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from datetime import datetime
from IPython.display import display, clear_output

# ===== 1. PROJECT SETTINGS =====

USERS = {
    "Vivek": "1234",
    "Rahul": "1111",
    "Aman": "2222",
    "Admin": "admin123"
}

# Simulated IPs for the college demo
USER_IPS = {
    "Vivek": "192.168.1.10",
    "Rahul": "192.168.1.25",
    "Aman": "192.168.1.15",
    "Admin": "192.168.1.50"
}

LOG_FILE = "login_logs.csv"
THRESHOLD = 3

# Create CSV file if it does not exist
if not os.path.exists(LOG_FILE):
    pd.DataFrame(
        columns=["Time", "User", "Status", "IP"]
    ).to_csv(LOG_FILE, index=False)

# ===== 2. LOGIN PAGE =====

title = widgets.HTML(
    value="<h2>Cybersecurity Log Analyzer</h2>"
          "<p>Educational Login Simulation</p>"
)

username = widgets.Text(
    description="Username:",
    placeholder="Enter username"
)

password = widgets.Password(
    description="Password:",
    placeholder="Enter password"
)

login_button = widgets.Button(
    description="LOGIN",
    button_style="primary"
)

analyze_button = widgets.Button(
    description="Analyze Logs",
    button_style="success"
)

report_button = widgets.Button(
    description="Download CSV Report"
)

output = widgets.Output()

# ===== 3. AUTOMATIC LOG GENERATION =====

def save_log(user, status, ip):
    new_log = pd.DataFrame([{
        "Time": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "User": user,
        "Status": status,
        "IP": ip
    }])

    new_log.to_csv(
        LOG_FILE,
        mode="a",
        header=False,
        index=False
    )

def login_system(button):
    with output:
        clear_output(wait=True)

        user = username.value.strip()
        pwd = password.value

        if not user:
            print("Please enter a username.")
            return

        # Demo-only IP mapping, not the visitor's actual IP
        ip = USER_IPS.get(user, "10.0.0.50")

        if user in USERS and USERS[user] == pwd:
            save_log(user, "LOGIN_SUCCESS", ip)
            print("LOGIN SUCCESSFUL")
        else:
            save_log(user, "LOGIN_FAILED", ip)
            print("LOGIN FAILED")

        print("User:", user)
        print("Demo IP:", ip)
        print("Time:", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
        print("Attempt automatically saved to", LOG_FILE)

        # Clear password field after attempt
        password.value = ""

# ===== 4. LOG ANALYSIS + SUSPICIOUS DETECTION =====

def analyze_logs(button=None):
    with output:
        clear_output(wait=True)

        df = pd.read_csv(LOG_FILE)

        if df.empty:
            print("No login attempts recorded yet.")
            return

        df["User"] = df["User"].fillna("Unknown").astype(str)
        df["IP"] = df["IP"].fillna("Unknown").astype(str)
        df["Status"] = df["Status"].fillna("").astype(str).str.strip()

        success = int((df["Status"] == "LOGIN_SUCCESS").sum())
        failed = int((df["Status"] == "LOGIN_FAILED").sum())

        failed_df = df[df["Status"] == "LOGIN_FAILED"]

        failed_ips = failed_df.groupby("IP").size()
        failed_users = failed_df.groupby("User").size()

        suspicious_ips = failed_ips[failed_ips >= THRESHOLD]
        suspicious_users = failed_users[failed_users >= THRESHOLD]

        print("===== SECURITY REPORT =====")
        print("Total login attempts:", len(df))
        print("Successful logins:", success)
        print("Failed logins:", failed)
        print("\n===== ALL LOGIN RECORDS =====")
        display(df.tail(20))

        print("\n===== SUSPICIOUS IP ADDRESSES =====")
        if suspicious_ips.empty:
            print("No IP reached the threshold.")
        else:
            display(suspicious_ips.rename("Failed Attempts").to_frame())

        print("\n===== SUSPICIOUS USERS =====")
        if suspicious_users.empty:
            print("No user reached the threshold.")
        else:
            display(suspicious_users.rename("Failed Attempts").to_frame())

        # Login activity graph
        plt.figure(figsize=(7, 4))
        plt.bar(
            ["Successful", "Failed"],
            [success, failed]
        )
        plt.title("Login Activity Analysis")
        plt.ylabel("Number of Attempts")
        plt.tight_layout()
        plt.show()

        # Failed attempts by IP graph
        if not failed_ips.empty:
            plt.figure(figsize=(8, 4))
            failed_ips.sort_values(ascending=False).plot(kind="bar")
            plt.title("Failed Login Attempts by IP")
            plt.xlabel("Demo IP Address")
            plt.ylabel("Failed Attempts")
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

def download_report(button):
    from google.colab import files
    files.download(LOG_FILE)

login_button.on_click(login_system)
analyze_button.on_click(analyze_logs)
report_button.on_click(download_report)

# ===== 5. DISPLAY COMPLETE PROJECT =====

display(
    title,
    username,
    password,
    login_button,
    analyze_button,
    report_button,
    output
)

print("Project ready! Use the login form above.")
print("Demo accounts: Vivek / 1234, Rahul / 1111,")
print("Aman / 2222, Admin / admin123")
```
